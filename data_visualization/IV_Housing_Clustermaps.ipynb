```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Copy of Copy of IVTU HOUSING DATA (updated) - 1993-2023.csv")
```

```python
df.head()
```

```python
df = df.drop(df[df['YEAR'] == 1993].index)
df = df.drop(df[df['YEAR'] == 1995].index)
df = df.drop(columns = ['# OF UNITS', 'STREET #', 'GAS (Y/N)', 'ELECTRIC (Y/N)', 'YARD (Y/N)', 'CABLE (Y/N)'], axis = 1)
df['MONTHLY RENT ($)'] = pd.to_numeric(df['MONTHLY RENT ($)'].str.replace(',','').str.replace('$',''), downcast = "float")
df['SECURITY ($)'] = pd.to_numeric(df['SECURITY ($)'].str.replace('$','').str.replace(',',''), downcast = "float")
df['BATHROOM'].replace('1+', '1', inplace = True)
df['PROPERTY TYPE'].replace('DUPLX', 'DUPLEX', inplace = True)
df['LEASOR'].replace('SFM VISTA DEL MAR', 'SFMVDM', inplace = True)
df.head()
```

```python
df_encoded = df
df_encoded['STREET NAME'], street_labels = pd.factorize(df_encoded['STREET NAME'])
df_encoded['UNIT #'], unit_labels = pd.factorize(df_encoded['UNIT #'])
df_encoded['PROPERTY TYPE'], property_labels = pd.factorize(df_encoded['PROPERTY TYPE'])
df_encoded['LEASOR'], leasor_labels = pd.factorize(df_encoded['LEASOR'])
df_encoded['FURNISHED (Y/P/N)'], furnished_labels = pd.factorize(df_encoded['FURNISHED (Y/P/N)'])
df_encoded['WATER (Y/N)'], water_labels = pd.factorize(df_encoded['WATER (Y/N)'])
df_encoded['TRASH (Y/N)'], trash_labels = pd.factorize(df_encoded['TRASH (Y/N)'])
df
```

```python
import pandas as pd
import numpy as np
import regex
import seaborn
```

```python
IV = pd.read_csv(r'C:\Users\Felix\Documents\Datasets_For_IV_Housing\ivtu_data_cleaned.csv')
IV.replace(r'^\s*$',np.nan,regex=True)
IV["PROPERTY TYPE"] = IV["PROPERTY TYPE"].map({'0':'NaN'})
```

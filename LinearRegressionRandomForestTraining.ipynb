```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
```

```python
model_df = pd.read_csv("CompletePreprocessedData.csv")
```

```python
X = model_df[['YEAR', 'STREET NAME', 'BEDROOM', 'BATHROOM', 'LEASOR']]
y = model_df[['MONTHLY RENT ($)']]
X_train, X_val, y_train, y_val = train_test_split(X, y, train_size = 0.8, test_size = 0.2)
```

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
lr_model.score(X_val, y_val)
```

```python
weights = lr_model.coef_
weights
```

```python
scores = np.zeros(100)
for i in range(100):
    X_train, X_val, y_train, y_val = train_test_split(X, y, train_size = 0.8, test_size = 0.2)
    lr_model = LinearRegression()
    lr_model.fit(X_train, y_train)
    scores[i] = lr_model.score(X_val, y_val)
scores.mean()
```

```python
for i in range(100):
    X_train, X_val, y_train, y_val = train_test_split(X, y, train_size = 0.8, test_size = 0.2)
    y_train = np.ravel(y_train)
    y_val = np.ravel(y_val)
    rf_model = RandomForestRegressor()
    rf_model.fit(X_train, y_train)
    scores[i] = rf_model.score(X_val, y_val)
scores.mean()
```

```python
params = {'n_estimators': [10, 50, 100, 500],
          'max_depth': [None, 10, 20, 30],
          'min_samples_split': [2, 5, 10, 20]}
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, y, train_size = 0.8, test_size = 0.2)
y_train = np.ravel(y_train)
y_val = np.ravel(y_val)
rf_model = RandomForestRegressor()
```

```python
gs = GridSearchCV(estimator = rf_model, param_grid = params, cv = 10, n_jobs = -1)
gs.fit(X_train, y_train)
best_params = gs.best_params_
best_model = gs.best_estimator_
```

```python
best_params
```

```python
best_model.score(X_val, y_val)
```

```python
for i in range(100):
    X_train, X_val, y_train, y_val = train_test_split(X, y, train_size = 0.8, test_size = 0.2)
    y_train = np.ravel(y_train)
    y_val = np.ravel(y_val)
    best_model.fit(X_train, y_train)
    scores[i] = best_model.score(X_val, y_val)
scores.mean()
```
